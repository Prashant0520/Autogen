# AutoGen Assistant with Memory

```python
from autogen_agentchat.agents import AssistantAgent
from autogen_agentchat.ui import Console
from autogen_core.memory import ListMemory, MemoryContent, MemoryMimeType
from autogen_ext.models.openai import OpenAIChatCompletionClient
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
# Add your API key here before running the assistant.
api_key = os.environ.get("AICREDITS_API_KEY")
base_url = os.environ.get("BASE_URL")

if not api_key:
    raise ValueError("Please set the OpenAI_API_KEY")

if not base_url:
    raise ValueError("Please set the BASE_URL")
```

```python
userMemory = ListMemory()
```

```python
await userMemory.add(MemoryContent(
    content='The weather unit should be degree celsius (metric)',
    mime_type=MemoryMimeType.TEXT
))

await userMemory.add(MemoryContent(
    content='The user is vegetarian',
    mime_type=MemoryMimeType.TEXT
))
```

```python
async def get_weather(city: str, units: str = "imperial") -> str:
    if units == "imperial":
        return f"The weather in {city} is 73 °F and Sunny."
    elif units == "metric":
        return f"The weather in {city} is 23 °C and Sunny."
    else:
        return f"Sorry, I don't know the weather in {city}."
```

```python
assistant_agent = AssistantAgent(
    'weather_assistant',
    model_client=OpenAIChatCompletionClient(model='gpt-4o', api_key=api_key, base_url= base_url,),
    tools=[get_weather],
    memory=[userMemory]
)
```

```python
stream = assistant_agent.run_stream(task='What is the weather in Delhi?')
await Console(stream)
```

```python
stream = assistant_agent.run_stream(task='Can you give me a good high-protein diet?')
await Console(stream)
```
