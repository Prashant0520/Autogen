```python
import os
from autogen_ext.models.openai import OpenAIChatCompletionClient
from autogen_core.models import ModelFamily
from autogen_agentchat.agents import AssistantAgent
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
# os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
# api_key=os.environ["OPENAI_API_KEY"]
os.environ["AICREDITS_API_KEY"]=os.getenv("AICREDITS_API_KEY")
api_key=os.environ["AICREDITS_API_KEY"]
os.environ["BASE_URL"]=os.getenv("BASE_URL")
base_url=os.environ["BASE_URL"]
os.environ["OPENROUTER_API_KEY"]=os.getenv("OPENROUTER_API_KEY")
openrouter_api_key=os.environ["OPENROUTER_API_KEY"]
```

```python
# model_client = OpenAIChatCompletionClient(model='gpt-4o', api_key=api_key)
model_client = OpenAIChatCompletionClient(model='gpt-5-nano', api_key=api_key, base_url=base_url)
```

```python
agent_1 = AssistantAgent(
    name="my_first_assistant",
    model_client=model_client
)

# Run a task
result = await agent_1.run(task="Tell me something about you?")
print(result)
```

```python
from autogen_core.models import UserMessage
from autogen_ext.models.openai import OpenAIChatCompletionClient

model_client = OpenAIChatCompletionClient(
    model="google/gemini-2.5-flash",
    api_key=api_key, 
    base_url=base_url,
    model_info={
        "vision": False,
        "function_calling": True,
        "json_output": False,
        "family": "gemini",
        "structured_output": False,
    },
)

response = await model_client.create([UserMessage(content="What is the capital of France?", source="user")])
print(response)
await model_client.close()
```

```python
response.content
```
